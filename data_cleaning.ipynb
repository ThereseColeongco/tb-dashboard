# Import libraries

```python
import pandas as pd
from pathlib import Path
```

# Read files in as dataframes

```python
main_dir = Path("data")
dfs = {}
for file in main_dir.rglob("*"):  # go into every sub-directory in main_dir
    # read every csv file except the data dictionary into a df and store the dfs in a dictionary where the key is the file name
    if file.is_file() and file.suffix.lower() == ".csv":
        df = pd.read_csv(file)
        dfs[file.stem] = df
list(dfs.keys())
```

# Check files for missing values and which columns

```python
for name, df in dfs.items():
    if "data_dictionary" not in name:
        print(f"Percentage of data missing in {name}")
        print(f"{((df.isna().mean() * 100).round(2).astype(str) + '%').to_string()}\n")
```
